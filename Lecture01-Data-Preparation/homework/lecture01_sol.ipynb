# Netflix TV Shows Analysis with Pandas

## **Giới thiệu bài tập:**

Netflix là một trong những nền tảng phát trực tuyến video và truyền thông hàng đầu thế giới. Tính đến giữa năm 2021, Netflix cung cấp hơn 8000 phim và chương trình truyền hình, thu hút hơn 200 triệu người dùng trên toàn cầu.

Trong bài tập này, chúng ta sẽ sử dụng dataset ***Netflix Movies and TV Shows*** (file ***netflix_titles.csv***) cùng với thư viện Pandas để thực hiện phân tích dữ liệu.

Dataset này chứa danh sách tất cả các phim và chương trình truyền hình có trên Netflix, đi kèm với các thông tin chi tiết như: diễn viên, đạo diễn, xếp hạng, năm phát hành, thời lượng, và nhiều thông tin khác.

Mỗi hàng trong dataset đại diện cho một chương trình hoặc bộ phim, với 12 cột chứa thông tin mô tả về nội dung đó:

1. **show_id**: Mã định danh của chương trình hoặc bộ phim trên Netflix.
2. **type**: Loại nội dung, như "Movie" (Phim) hoặc "TV Show" (Chương trình truyền hình).
3. **title**: Tên của chương trình hoặc bộ phim.
4. **director**: Tên đạo diễn của chương trình hoặc bộ phim.
5. **cast**: Danh sách các diễn viên tham gia.
6. **country**: Quốc gia sản xuất chương trình hoặc bộ phim.
7. **date_added**: Ngày nội dung được phát hành trên Netflix.
8. **release_year**: Năm phát hành của chương trình hoặc bộ phim.
9. **rating**: Xếp hạng theo hệ thống như PG, R, v.v.
10. **duration**: Thời lượng của phim hoặc số mùa của chương trình truyền hình.
11. **listed_in**: Thể loại của chương trình hoặc bộ phim.
12. **description**: Mô tả ngắn gọn về nội dung.


## Load và tìm hiểu thông tin dữ liệu

```python
# import thư viện cần thiết
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dữ liệu từ file netflix_titles.csv và DataFrame df
df = pd.read_csv('netflix_titles.csv')

# hiển thị vài dòng đầu
df.head()
```

```python
# Run this cell to test your result
assert df.shape == (8807, 12)
```

```python
# mô tả thông tin dữ liệu
df.info()
```

**Giải thích kết quả**: df có tổng cộng 8807 dòng, 12 cột (8807 entries và mỗi entries có 12 features)

**Các cột có giá trị bị thiếu (missing values), số lượng non-null nhỏ hơn 8807:**
* director
* cast
* country
* date-added
* rating
* duration

## Số lượng missing values ở tất cả các cột trong dataset

```python
# Số lượng missing values ở tất cả các cột trong dataset
missing_values = df.isna().sum()
missing_values
```

```python
# Run this cell to test your result
assert missing_values.sum() == 4307
```

## So sánh tỉ lệ phim (Movies) và chương trình truyền hình (TV Shows)

```python
#tìm tổng số lượng của từng loại chương trình trong cột type
show_type_count = df['type'].value_counts()
show_type_count
```

```python
# Tính tỉ lệ phần trăm của từng loại chương trình
# cách 1:
show_type_percentage = df['type'].value_counts(normalize=True) * 100

# cách 2:
# show_type_percentage = (show_type_count * 100)/len(df)

show_type_percentage
```

```python
# Run this cell to test your result
assert np.abs(show_type_percentage['Movie'] -69.615) < 0.01
```

```python
# Chạy cell này sau khi tìm được show_type_percentage để xem visualization

plt.figure(figsize=(4,4))
plt.pie(show_type_percentage, labels = show_type_percentage.index, autopct = '%.1f')
plt.show()
```

## Tìm top 10 quốc gia có nhiều nội dung trên Netflix nhất

```python
top_10_countries = df['country'].value_counts().head(10)
top_10_countries
```

```python
# Run this cell to test your result
assert set(top_10_countries.keys()) == set(['United States', 'India', 'United Kingdom', 'Japan', 'South Korea',
       'Canada', 'Spain', 'France', 'Mexico', 'Egypt'])
```

```python
# Chạy cell này sau khi tìm được top_10_countries để xem visualization
plt.figure(figsize=(10, 6))
sns.barplot(x=top_10_countries.values, y=top_10_countries.index)
plt.title('Top 10 quốc gia theo số lượng nội dung trên Netflix')
plt.xlabel('Số lượng')
plt.ylabel('Quốc gia')
plt.show()
```

## Tìm top 10 đạo diễn

```python
top_10_directors = df['director'].value_counts().head(10)
top_10_directors
```

```python
# Run this cell to test your result
assert set(top_10_directors.keys()) == set(['Rajiv Chilaka', 'Raúl Campos, Jan Suter', 'Suhas Kadav',
       'Marcus Raboy', 'Jay Karas', 'Cathy Garcia-Molina', 'Martin Scorsese',
       'Youssef Chahine', 'Jay Chapman', 'Steven Spielberg'])
```

```python
# Chạy cell này sau khi tìm được top_10_directors để xem visualization

plt.figure(figsize=(8, 6))
sns.barplot(x=top_10_directors.values, y=top_10_directors.index)
plt.title('Top 10 Đạo diễn trên nền tảng Netflix')
plt.xlabel('Số lượng chương trình')
plt.ylabel('Đạo diễn')
plt.show()
```

## Quốc gia nào có nhiều phim xếp hạng R-rated nhất?

```python
# Tìm số lượng các bộ phim R-rated theo từng quốc gia và sắp xếp theo thứ tự giảm dần
R_rated = df[df.rating == 'R']['country'].value_counts()#.sort_values(ascending=False)
R_rated
```

```python
# Run this cell to test your result
assert R_rated.shape == (182,)
```

```python
# Chạy cell này sau khi tìm được R_rated để xem visualization
top_10_R_rated = R_rated.head(10)
plt.figure(figsize=(8, 6))
sns.barplot(x=top_10_R_rated.values, y=top_10_R_rated.index)
plt.title('Top 10 quốc gia sản xuất nhiều phim nhãn R nhất')
plt.xlabel('Số lượng')
plt.ylabel('Quốc gia')
plt.show()
```
