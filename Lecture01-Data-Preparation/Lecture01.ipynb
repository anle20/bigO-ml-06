## List trong python

### Truy xuất dữ liệu list trong python

````python
arr = [3, 2, 10, -10, 4]

print(arr[3])
print(arr[1:4])
print(arr[-1:-4])
print(arr[-1:-4:-1])
print(arr[::-1])
````

### Gán giá trị giữa các list trong python

Việc gán bình thường chỉ là gán giá trị cho reference tới biến đó. Do vậy, thay đổi giá trị biến mới sẽ làm thay đổi giá trị biến cũ.

````python
arr = [1, 2, 3]
b = arr
b[1] = 5
print(arr)
print(b)
````

````python
arr = [1, 2, 3]
b = arr[:]
b[1] = 5
print(arr)
print(b)
````

Sử dụng indexing để copy và gán giá trị của một list.

````python
arr = [0, 1, [1, 2, 3]]
b = arr[:]
b[2].append(10)
print(arr)
print(b)
````

Tuy nhiên việc sử dụng ``[:]`` chỉ shallow copy đối với những phần tử trong mảng. Để thật sự deepcopy, ta sử dụng ``copy.deepcopy``.

````python
import copy

arr = [0, 1, [1, 2, 3]]
b = copy.deepcopy(arr)
b[2].append(10)
print(arr)
print(b)
````

### Các phép toán được cung cấp cho list

````python
arr = [1, 2, 3]
b = [4, 5, 6]
print(arr + b)
print(arr * 2)
try:
    print(arr - b)
except Exception as e:
    print('Can not perform arr - b:', e)

try:
    print(arr * b)
except Exception as e:
    print('Can not perform arr * b:', e)
````

## Tuple

Tuple là kiểu giá trị tương tự List. Tuy nhiên ta không thể thay đổi giá trị của một Tuple. Nói cách khác,Tuple là kiểu dữ liệu immutable.

````python
arr = (1, 2, 3)
b = 1, 2
c = (1, 2), (4, 5)
print(arr)
print(b)
print(c)
````

Không thể thay đổi giá trị của một Tuple.

````python
arr = (1, 2)
try: 
    arr[0] = 3
except Exception as e: 
    print(e.__class__.__name__, e)
````

## Set

Set chỉ lưu những phần tử có giá trị khác nhau.

````python
arr = {1, 2, 3, 1, 2, 3}
print(arr)
````

Có thể dùng set để đếm số lượng những phần tử khác nhau.

````python
a1 = [1, 2, 3, 2, 5, 1]
arr = set(a1)
print(len(arr))
````

### Các phép toán tập hợp trên Set

````python
green = {'an', 'binh', 'cuong', 'minh'}
blue = {'an', 'hung', 'hai'}
print(green.union(blue))
print(green.intersection(blue))
print(green.difference(blue))
print(green.symmetric_difference(blue))
````

````python
print(green | blue)
print(green & blue)
print(green - blue)
print(green ^ blue)
````

````python
titles = ['Data Scientist', 'Data Analysis', 'Software Engineer', 'Software Tester', 'DevOps Engineer', 'DevOps Architecture']
team_group = {}
for title in titles: 
    team = title.split()[0]
    if team not in team_group:
        team_group[team] = [title]
    else:
        team_group[team].append(title)
team_group
````

````python
titles = ['Data Scientist', 'Data Analysis', 'Software Engineer', 'Software Tester', 'DevOps Engineer', 'DevOps Architecture']
team_group = {}
for title in titles: 
    team = title.split()[0]
    team_group.setdefault(team, []).append(title)
team_group
````

## List, Set, Dict comprehension

Cách làm truyền thống sử dụng ``for``

````python
arr = list(range(1, 5))
for i in range(len(arr)): 
    arr[i] = arr[i] ** 2
print(arr)
````

Cách làm với list comprehension

````python
arr = list(range(1, 5))
arr = [x**2 for x in arr]
print(arr)
````

List comprehension với điều kiện ``if``

````python
arr = list(range(1, 10))
arr = [x**2 for x in arr if (x % 2 == 0)]
print(arr)
````
